```python
import requests

url = "https://query.wikidata.org/sparql"
query = """
SELECT ?lemma WHERE {
  ?lexeme dct:language wd:Q9035 ;
          wikibase:lemma ?lemma .
}
"""

response = requests.get(url, params={
    "query": query,
    "format": "json"
}, headers={"User-Agent": "MyApp/1.0"})

data = response.json()
words = sorted(set(
    r["lemma"]["value"]
    for r in data["results"]["bindings"]
))
```

```python
# remove words that include anything other than letters. This includes spaces, numbers, signs, etc.
words = sorted(set(
    w.lower() for w in words
    if all(c.isalpha() for c in w)
))
```

```python
with open("danish_words.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(words))

print(f"Saved {len(words)} unique Danish words to danish_words.txt")
```
